```python
X=np.arange(8)
Y=X+0.5
C=1.0/np.subtract.outer(X,Y)#构造柯西矩阵，计算X和Y的外积差，生成一个8*8的矩阵
print(np.linalg.det(C))#计算行列式
```

```python
for dtype in [np.int8,np.int32,np.int64]:
    print(np.iinfo(dtype).min,np.iinfo(dtype).max)
#np.iinfo(dtype)返回该整数类型的信息对象 ，int8：八位有符号整数，范围到-128到127

for dtype in [np.float32,np.float64]:
    print(np.finfo(dtype).min,np.finfo(dtype).max,np.finfo(dtype).eps)
```

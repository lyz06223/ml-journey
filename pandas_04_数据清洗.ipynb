```python
import pandas as pd
```

```python
df=pd.read_csv("data/titanic.csv")
```

```python
df.shape
df.head()
df.info()
```

```python
df.isnull().sum()#每列缺失多少个,逐个判断每个格子是不是空的，返回True/False  .sum()计算有多少个True
df.isnull().sum()/len(df)*100#换算成百分比  
```

```python
df=df.dropna(subset=["Embarked"])#subset只在Embarked为空时删行，其他列不动   
```

```python
df["Age"]=df.groupby(["Pclass","Sex"])["Age"].transform(lambda s:s.fillna(s.median()))
#df.groupby(["Pclass","Sex"] 按舱位和性别分成6组
#["Age"]只看Age这一列
#.transform() 对每一组分别处理，结果放回原来的位置
#lambda s:s.fillna(s.median()) 这一组里面，空的用本组中位数填 fillna()填数 s.median()中位数
```

```python
df["HasCabin"]=df["Cabin"].notnull().astype(int)#notnull不是空的是返回True和False  astype转型将true转化为1 false转化为0
df=df.drop(columns=["Cabin"])#drop删列
```

```python
df.groupby("HasCabin")["Survived"].agg(["mean","count"])#groupby 按HasCabin分堆，只看Survived
#agg是聚合可以一次算两个mean是算平均数count是计算个数
```

```python
#不要覆盖原始数据
df=pd.read_csv("data/titanic.csv")
df_clean=df.copy()
df_clean=df_clean.dropna(subset=["Embarked"])
df_clean["Age"]=df_clean.groupby(["Pclass","Sex"])["Age"].transform(lambda s:s.fillna(s.median()))
df_clean["HasCabin"]=df_clean["Cabin"].notnull().astype(int)
df_clean=df_clean.drop(columns=["Cabin"])
df_clean.groupby("HasCabin")["Survived"].agg(["mean","count"])
df_clean.to_csv("data/titanic_clean.csv",index=False)
```

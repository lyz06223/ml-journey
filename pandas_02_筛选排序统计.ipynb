```python
gdp = "地区生产总值(亿元)"

print(f"平均值  : {df[gdp].mean():>12,.2f} 亿元")
print(f"中位数  : {df[gdp].median():>12,.2f} 亿元")
print(f"最大值  : {df[gdp].max():>12,.2f} 亿元")
print(f"最小值  : {df[gdp].min():>12,.2f} 亿元")
print(f"标准差  : {df[gdp].std():>12,.2f} 亿元")
print(f"总和    : {df[gdp].sum():>12,.2f} 亿元")
```

## 分析结论

**数据**：全国 31 个省份，7 项经济与民生指标

### 1. 经济总量高度集中在少数省份

GDP 超过 5 万亿元的省份有 **8 个**（江苏、浙江、福建、山东、河南、湖北、广东、四川），约占全国的 1/4。

最高的是 **广东省（129118.6 亿元）**，最低的是 **西藏自治区（2132.6 亿元）**，**相差约 60 倍**。

### 2. 平均值明显高于中位数 —— 数据右偏

| 指标 | 数值 |
| --- | --- |
| 平均值 | 38821.36 亿元 |
| 中位数 | 29129.00 亿元 |

**平均值比中位数高出近 1 万亿**，说明少数经济大省把平均水平拉高了。**实际上有一半以上的省份，GDP 低于平均值。**

> **这就是为什么看数据不能只看平均值** —— 中位数才能反映"典型省份"的真实水平。

### 3. 房价最高的不是经济最强的

| 排名 | 省份 | 房价（元/㎡） | GDP（亿元） |
| --- | --- | --- | --- |
| 1 | 上海 | 40302 | 44652.8 |
| 2 | 北京 | 38240 | 41610.9 |
| 3 | 浙江 | 18576 | 77715.4 |

**浙江的 GDP 远高于上海和北京，但房价只有它们的一半左右。**

**结论**：房价不只由经济水平决定，还受**人口密度、土地供应、城市定位、政策调控**等因素影响。
